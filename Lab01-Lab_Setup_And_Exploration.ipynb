## Ensure proper permissions

Snowflake CoCo is capable of performing many tasks within your Snowflake account. However, it is limited by the role that it is using. Snowflake CoCo always has the permissions of whichever role is selected as the Default role.

To ensure that CoCo will have permissions to perform the tasks in the labs, set your default role to ACCOUNTADMIN:

Click your name in the bottom-left of the user interface.

Click your current role (under the label: **Switch Role**).

Hover the mouse over ACCOUNTADMIN and click the **Set as Default** button.

## Setup a Warehouse

Snowflake uses Warehouses to provide compute for most queries.

To create a Warehouse for these labs, click the Run button ( ![](resources/img/run_cell_button.png) ) in the top corner of the cell below.

```
%%sql -r dataframe_2
CREATE WAREHOUSE IF NOT EXISTS COCOLABS_WH WAREHOUSE_SIZE = XSMALL AUTO_SUSPEND = 60 INITIALLY_SUSPENDED = TRUE;
USE WAREHOUSE COCOLABS_WH;
SET my_user = CURRENT_USER();
ALTER USER IDENTIFIER($my_user) SET DEFAULT_WAREHOUSE = 'COCOLABS_WH';
```

## Explore Snowflake CoCo

Take a look to the right side of the user interface. You likely see a chat panel asking you how it can help.

If you don't see this panel, click the Snowflake CoCo icon in the bottom right:

<div style="max-width:400px;">

![](resources/img/open-cocome.png)</div>

### Ask Coco What it can do

Let's start simple. Type the following prompt into the chat panel and submit it.

<code style="display:block; user-select:all; padding: 15px;">
Show me what CoCo can do.
</code>

CoCo will explain its capabilities and will likely give you a few other sample prompts.

### Try a different model

Snowflake CoCo has access to a few different LLMs. Different models may give different results. 

At the bottom of the chat panel, click the **Auto** dropdown. You will see a list of available LLMs. The models listed here are updated as new versions are published.

Select a model other than Auto, then repeat the prompt. (Hint: you can press the Up Arrow key to go through a history of your prompts.)

Select a third model and repeat the prompt.

Notice that the information presented is similar, but each model presents it a bit differently. Each model has its own strengths, especially when it comes to generating code. To simplify things, **Auto** mode will automatically select the model that is best for your prompt.

When you have finished, select **Auto** again.

### Start a new chat

Snowflake CoCo will remember your conversation and sometimes refer to things you typed previously. This is often very helpful, but can be undesired if you are starting a new workflow.

At the top of the chat panel, click the **New chat** button ( ![](resources/img/newchatbutton.png) ).

You now have a new chat.

Enter the following prompt:

<code style="display:block; user-select:all; padding: 15px;">
What databases do I have access to?
</code>

CoCo will show that databases that its role (ACCOUNTADMIN) has access to.

### Chat History

You can always return to a previous chat by viewing the chat history.

At the top of the chat panel, click the **Chat history** button ( ![](resources/img/chathistorybutton.png) ).

You will see your previous chat. Oftentimes, these chats will be automatically labelled with a summary of their content.

Click the entry for your previous chat to return to it.

## Managing Costs

It can be easy to get carried away with Snowflake CoCo and use many credits. Fortunately, it is possible to set a daily limit.

These labs are expected to use no more than 20 credits.

### Limiting Usage

If you want to prevent yourself from using too many credits, click the Run button ( ![](resources/img/run_cell_button.png) ) in the top corner of the cell below. It will limit your daily CoCo usage to 20 credits.

```
%%sql -r dataframe_1
SET my_user = CURRENT_USER();
ALTER USER IDENTIFIER($my_user) SET CORTEX_CODE_SNOWSIGHT_DAILY_EST_CREDIT_LIMIT_PER_USER = 20;
```

### Viewing Cost Consumption

You can see how much credits you have used with Snowflake CoCo in the Snowsight GUI.

In the menu on the left, click **Admin**. Then, right-click **Cost management** and select **Open in a New Tab**. This way, you can still follow the instructions documented here.

<div style="max-width:400px;">

![](resources/img/open-cost-management.png)</div>

In the **Cost Management** page, click the **Consumption** tab.

From the **Service Type** menu, select **CoCo Snowsight**.

If you are asked to select a warehouse, select **COCOLABS_WH**.

This shows how many credits have been used by CoCo in Snowsight over the selected time period. Keep an eye on this metric if you are concerned about using up your free credits.

<div style="max-width:600px;">

![](resources/img/cococost.png)</div>

### Viewing personal CoCo usage

The Cost Management page shows overall usage across all users of the account. Since you are the only user of your trial account, it will only show your usage. But if you had multiple users on your account, you might want to see usage for them individually.

Snowflake does provide views to access this information. But it is easier to just ask CoCo. Enter the following prompt into the CoCo chat panel:

<code style="display:block; user-select:all; padding: 15px;">
How can I determine the cost of my own CoCo usage?
</code>

The information shown may not be up to date. But you can use it to track usage by individual users.
