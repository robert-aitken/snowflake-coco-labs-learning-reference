## Create Skeletal Streamlit App

You will use Streamlit in Snowflake to build the data application. Streamlit provides a rich library for creating interactive visualizations and was created specifically for building data applications.

You do not need to know how to write Streamlit code. Snowflake CoCo will do that for you.

### Create app with three pages

Your dashboard will show three types of data: Team Information, Player Statistics, and Salary Data.

Streamlit is capable of creating an app with multiple pages, with a navigation menu on the left side to navigate between the pages. Let's start by creating just this menu. For now, each menu item will just show a placeholder page.

Enter the following prompt:

<code style="display:block; user-select:all; padding: 15px;">
Create a new Streamlit app in this Workspace called "PFL Dashboard". The app has a navigation menu with three items: Team Information, Player Statistics, and Salary Data. The items link to placeholder pages that just show a title.
</code>
